```python
# Cell 4: Interactive Concept Query & Hyper-Edge Inspector
def inspect_concept(concept_name):
    if concept_name in G:
        data = G.nodes[concept_name]
        print(f"=== {concept_name} ===")
        print(f"Category: {data.get('category')}")
        print(f"Symbol / Formula: {data.get('symbol')}")
        print(f"Value / Mechanism: {data.get('val')}")
        print("\n--- Outgoing Relations ---")
        for _, dst, edata in G.out_edges(concept_name, data=True):
            print(f"  → {dst} [{edata.get('label')}]")
        print("\n--- Incoming Relations ---")
        for src, _, edata in G.in_edges(concept_name, data=True):
            print(f"  ← {src} [{edata.get('label')}]")
    else:
        print(f"Concept '{concept_name}' not found. Available concepts:")
        print(list(G.nodes()))

# Example inspection:
inspect_concept("Master Cost Function C(X)")
```
